```python
# Diccionario vacío
plomeros = {}


# FUNCION AGREGAR
def agregar_plomero():

    nombre = input("Ingrese nombre del plomero: ")
    kit = input("Ingrese kit de herramientas: ")
    emergencia = input("Disponibilidad de emergencia: ")
    materiales = input("Materiales frecuentes: ")
    licencia = input("Licencias: ")
    calificacion = input("Calificación de clientes: ")
    anios = input("Años de servicio: ")

    plomeros[nombre] = {
        "Kit de herramientas": kit,
        "Disponibilidad de emergencia": emergencia,
        "Materiales frecuentes": materiales,
        "Licencias": licencia,
        "Calificación de clientes": calificacion,
        "Años de servicio": anios
    }

    print("\nPlomero agregado correctamente")


# FUNCION LISTAR
def listar_plomeros():

    if len(plomeros) == 0:
        print("\nNo hay plomeros registrados")

    else:

        print("\n================================ TABLA DE PLOMEROS =================================")

        # ENCABEZADOS
        print(
            "{:<20} {:<30} {:<15} {:<20} {:<15} {:<15} {:<10}".format(
                "Nombre",
                "Herramientas",
                "Emergencia",
                "Materiales",
                "Licencias",
                "Calificacion",
                "Años"
            )
        )

        print("-" * 131)

        # FOR CLAVE Y VALOR
        for clave, valor in plomeros.items():

            print(
                "{:<20} {:<30} {:<15} {:<20} {:<15} {:<15} {:<10}".format(
                    clave,
                    valor["Kit de herramientas"],
                    valor["Disponibilidad de emergencia"],
                    valor["Materiales frecuentes"],
                    valor["Licencias"],
                    str(valor["Calificación de clientes"]),
                    str(valor["Años de servicio"])
                )
            )


# FUNCION BUSCAR
def buscar_plomero():

    nombre = input("\nIngrese nombre a buscar: ")

    if nombre in plomeros:

        print("\nPlomero encontrado\n")

        for key, value in plomeros[nombre].items():
            print(f"{key}: {value}")

    else:
        print("El plomero no existe")


# FUNCION BORRAR
def borrar_plomero():

    nombre = input("\nIngrese nombre a borrar: ")

    if nombre in plomeros:
        del plomeros[nombre]
        print("Plomero eliminado")

    else:
        print("El plomero no existe")


# FUNCION ACTUALIZAR
def actualizar_plomero():

    nombre = input("\nIngrese nombre a actualizar: ")

    if nombre in plomeros:

        kit = input("Nuevo kit de herramientas: ")
        emergencia = input("Nueva disponibilidad de emergencia: ")
        materiales = input("Nuevos materiales frecuentes: ")
        licencia = input("Nuevas licencias: ")
        calificacion = input("Nueva calificación: ")
        anios = input("Nuevos años de servicio: ")

        plomeros[nombre] = {
            "Kit de herramientas": kit,
            "Disponibilidad de emergencia": emergencia,
            "Materiales frecuentes": materiales,
            "Licencias": licencia,
            "Calificación de clientes": calificacion,
            "Años de servicio": anios
        }

        print("Plomero actualizado correctamente")

    else:
        print("El plomero no existe")


# MENU PRINCIPAL
while True:

    print("\n=========== MENU ===========")
    print("1. Agregar plomero")
    print("2. Listar plomeros")
    print("3. Buscar plomero")
    print("4. Borrar plomero")
    print("5. Actualizar plomero")
    print("6. Salir")

    opcion = input("Seleccione una opcion: ")

    if opcion == "1":
        agregar_plomero()

    elif opcion == "2":
        listar_plomeros()

    elif opcion == "3":
        buscar_plomero()

    elif opcion == "4":
        borrar_plomero()

    elif opcion == "5":
        actualizar_plomero()

    elif opcion == "6":
        print("Programa finalizado")
        break

    else:
        print("Opcion incorrecta")
```
